```python
def get_locations_at_shot(df, column_name):
    at_shot_locations =[]
    for server_location_values in df[column_name].tolist():
        float_values = [float(x) for x in server_location_values.strip('][').split(', ')]
        if len(float_values) > 0:
            at_shot_locations.append(float_values[0])
        else:
            at_shot_locations.append(np.nan)
    return at_shot_locations
```

```python
def define_neutral_shots(df, shot_no_min = 4, player_location_wide_max = 5, player_movement_speed_max = 3, player_acc_max = 5, player_distance_moved_before_contact_max = 5, player_dist_from_the_net_min = 10, 
                         player_dist_from_the_net_max = 17, incoming_ball_speed_before_contact_max=55, incoming_ball_speed_before_contact_min=20, previous_shot_speed_max = 150, previous_shot_speed_min = 60, 
                         rebound_depth_max = 11.6):
    
    #limit shot number
    df_neutral_shot_no = df[df['shot_no'] >= shot_no_min].copy()
    
    #neither player should be standing outside of the 5m from the middle line when the ball is hit
    df_neutral_not_too_wide = df_neutral_shot_no[(df_neutral_shot_no['sel_player_location_at_shot_y_mirrored'] <= player_location_wide_max)&(df_neutral_shot_no['opponent_location_at_shot_y_mirrored'] <= player_location_wide_max) &
                      (df_neutral_shot_no['sel_player_location_at_shot_y_mirrored'] >= -(player_location_wide_max))&(df_neutral_shot_no['opponent_location_at_shot_y_mirrored'] >= -(player_location_wide_max))]

    #Average speed of movement before the shot was less than X
    df_neutral_player_speed = df_neutral_shot_no[(df_neutral_shot_no['sel_player_avg_speed_previous'] <= player_movement_speed_max) & (df_neutral_shot_no['opponent_avg_speed_previous'] <= player_movement_speed_max)]
    
    #ACCELERATIONS SHOULD STILL BE TUNED
    #Average acceleration of movement before the shot was less than X (should be changed to average maybe)
    df_neutral_player_accelerations = df_neutral_shot_no[(df_neutral_shot_no['sel_player_accelerations_window_max_previous'] <= player_acc_max) & (df_neutral_shot_no['opponent_accelerations_window_max_previous'] <= player_acc_max)]

    #Average deceleration of movement before the shot was less than X (should be changed to average maybe)
    #df_neutral_player_deceleration = df_neutral_shot_no[(df_neutral_shot_no['sel_player_decelerations_window_max_previous'] >= -(sel_player_acc)) & (df_neutral_shot_no['opponent_decelerations_window_max_previous'] >= -(sel_player_acc))]

    #Distance a player makes before the contact
    df_neutral_player_distance_moved = df_neutral_shot_no[(df_neutral_shot_no['sel_player_shot_distance_moved_previous'] <= player_distance_moved_before_contact_max) & (df_neutral_shot_no['opponent_shot_distance_moved_previous'] <= player_distance_moved_before_contact_max)]

    #Player position not too much in the court
    df_neutral_player_too_close = df_neutral_shot_no[(df_neutral_shot_no['sel_player_location_at_shot_x_abs'] >= player_dist_from_the_net_min)&(df_neutral_shot_no['opponent_location_at_shot_x_abs'] >= player_dist_from_the_net_min)]

    #Player position not too far from the net
    df_neutral_player_too_far = df_neutral_shot_no[(df_neutral_shot_no['sel_player_location_at_shot_x_abs'] < player_dist_from_the_net_max)&(df_neutral_shot_no['opponent_location_at_shot_x_abs'] < player_dist_from_the_net_max)]

    #Speed of the incomming shot
    df_neutral_incomming_ball_speed = df_neutral_shot_no[(df_neutral_shot_no['speed_after_bounce_before_hit_previous'] <= incoming_ball_speed_before_contact_max) &(df_neutral_shot_no['speed_after_bounce_before_hit_previous'] >= incoming_ball_speed_before_contact_min)]
    df_neutral_shot_speed = df_neutral_shot_no[(df_neutral_shot_no['previous_shot_speed'] <= previous_shot_speed_max)&(df_neutral_shot_no['previous_shot_speed'] >= previous_shot_speed_min)]

    #If incomming ball bounces (very) close to the baseline it is not neutral ball
    df_neutral_bounce_too_deep = df_neutral_shot_no[df_neutral_shot_no['REBOUND_X_abs_previous'] < rebound_depth_max]

    df_neutral_situation_shots = df[df.index.isin(df_neutral_shot_no.index) & df.index.isin(df_neutral_not_too_wide.index) & df.index.isin(df_neutral_player_speed.index)
                                   & df.index.isin(df_neutral_player_distance_moved.index) & df.index.isin(df_neutral_player_too_close.index) & df.index.isin(df_neutral_player_too_far.index)
                                   & df.index.isin(df_neutral_incomming_ball_speed.index) & df.index.isin(df_neutral_player_accelerations.index) & df.index.isin(df_neutral_shot_speed.index) 
                                   & df.index.isin(df_neutral_bounce_too_deep.index)].copy()
    
    return df_neutral_situation_shots
def process_df(a, selected_player_name):
    a['opponent_shot_distance_moved'] = np.where(a['server_name'] != selected_player_name, a['server_shot_distance_moved'], a['returner_shot_distance_moved'])

    a['sel_player_location_at_shot_y_mirrored']  = np.where(a['sel_player_location_at_shot_x'] > 0, -1*a['sel_player_location_at_shot_y'], a['sel_player_location_at_shot_y'])
    a['sel_player_avg_speed_previous'] = a['sel_player_avg_speed'].shift(1)
    a['opponent_avg_speed'] = np.where(a['server_name'] == selected_player_name, a['returner_speeds_avg'], a['server_speeds_avg'])
    a['opponent_avg_speed_previous'] = a['opponent_avg_speed'].shift(1)
    a['sel_player_decelerations_window_max'] = np.where(a['server_name'] == selected_player_name, a['server_decelerations_window_max'], a['returner_decelerations_window_max'])
    a['sel_player_accelerations_window_max'] = np.where(a['server_name'] == selected_player_name, a['server_accelerations_window_max'], a['returner_accelerations_window_max'])
    a['sel_player_decelerations_window_max_previous'] = a['sel_player_decelerations_window_max'].shift(1)
    a['sel_player_accelerations_window_max_previous'] = a['sel_player_accelerations_window_max'].shift(1)
    a['opponent_accelerations_window_max'] = np.where(a['server_name'] == selected_player_name, a['returner_accelerations_window_max'], a['server_accelerations_window_max'])
    a['opponent_decelerations_window_max'] = np.where(a['server_name'] == selected_player_name, a['returner_decelerations_window_max'], a['server_decelerations_window_max'])
    a['opponent_accelerations_window_max_previous'] = a['opponent_accelerations_window_max'].shift(1)
    a['opponent_decelerations_window_max_previous'] = a['opponent_decelerations_window_max'].shift(1)
    a['opponent_shot_distance_moved_previous'] = a['opponent_shot_distance_moved'].shift(1)
    a['speed_after_bounce_before_hit_previous'] = a['speed_after_bounce_before_hit'].shift(1)
    df_neutral_shots = define_neutral_shots(a)
    a['is_neutral'] = 0
    a.loc[a.shot_id.isin(df_neutral_shots.shot_id), 'is_neutral'] = 1
    return a
```

```python
import pandas as pd
l = pd.read_parquet('/Users/marinkreso/Documents/gsa_post_match/nicegui_dir/new_code4/leaderboard_haddad_new.parquet')
import re
data = []
clay_tournaments = ['Rome', 'Strasbourg', 'RG', 'Madrid', 'Monte Carlo', 'Monte_Carlo', 'Munich', 'Barcelona', 'Umag', 'Kitzbuhel', 'Bastad', 'Gstaad', 'Geneva', 'Monte-Carlo', 'Hamburg', 'Buenos_Aires', 'Lyon', 'Rio',  
                    'Santiago', 'Buenos_Aires', 'Houston', 'Marrakech', 'Bucharest', 'Charleston', 'Hamburg', 'Barcelona', 'Munich']
grass_tournaments = ['Wimbledon', 'Hertogenbosh', 'Halle', 'Queens', 'Stuttgart', 'Malorca', 'Mallorca', 'Hertogenbosch', 'Eastbourne', 'Berlin']
clay_tournaments = [x.upper() for x in clay_tournaments]
grass_tournaments = [x.upper() for x in grass_tournaments]
#forbidden = [
#        'AO_2025_1st_Djokovic_Basavareddy',
#        'AO_2025_SF_Djokovic_Zverev',
#        'AO_2025_QF_Djokovic_Alcaraz',
#        'AO_2025_R16_Djokovic_Lehecka',
#        'AO_2025_R32_Djokovic_Machac'
#    ]
forbidden = ['Indian_Wells_2023_R16_Medvedev_Zverev',
       'Indian_Wells_2023_R32_Ruusuvuori_Zverev',
       'Indian_Wells_2023_R64_Cachin_Zverev',
                'Monte_Carlo_2023_R16_Medvedev_Zverev',
       'Monte_Carlo_2023_R32_Bautista-Agut_Zverev',
       'Monte_Carlo_2023_R64_Bublik_Zverev']
l = l[~l.match_id.isin(forbidden)]
for _, row in l[l.match_id != 'ALL'][['match_id', 'player_name', 'opponent_name', 'surface']].drop_duplicates().iterrows():
    #if not row['match_id'] in df_all.match_id.unique():
    #    continue
    if 'combined' in row['match_id']:
        continue
    pattern = re.compile(r'(.+?)_(\d{4})_(.+)')
    # Use the pattern to match and extract the partssss
    matches = pattern.match(row['match_id'])
    parts = matches.groups()

    tournament = parts[0].replace('_', ' ').strip().upper()
    if tournament == 'MONTREAL' or tournament == 'TORONTO':
        tournament = 'ROGERS CUP'
    year = parts[1]
    
    if tournament in grass_tournaments and row['surface'] != 'grass':
        row['surface'] = 'grass'
        if tournament.lower() == 'stuttgart' and ('swiatek' in row['match_id'].lower() or  'rybakina' in row['match_id'].lower()) :
            row['surface'] = 'clay'
            

        #print(row['match_id'], 'grass')
    if tournament in clay_tournaments and row['surface'] != 'clay':
        row['surface'] = 'clay'
        #print(row['match_id'], 'clay')
    if tournament == 'RG':
        tournament = 'ROLAND GARROS'
    if tournament == 'AO':
        tournament = 'AUSTRALIAN OPEN'
    if row['player_name'] == 'DJOKOVIC' and year != '2025':
        continue
    
    try:
        round_number, _, _ = parts[2].replace('Haddad_Maia', 'Haddad-Maia').replace('Carreno_Busta', 'Carreno-Busta').replace('Bautista_Agut', 'Bautista-Agut').replace('van_de_Zandschulp', 'van-de-Zandschulp').replace('Davidovich_Fokina', 'Davidovich-Fokina').split('_')
        data.append({
            'PLAYER': row['player_name'],
            'OPPONENT': row['opponent_name'],
            'TOURNAMENT': tournament.replace(' ', '-').replace('_', '-'),
            'YEAR': year,
            'ROUND': round_number,
            'SURFACE': row['surface'],
            'FORMAT': 'HE',
            'match_id': row['match_id'],
        })
    except:
        print(parts)
        continue
import json
with open('/Users/marinkreso/Documents/gsa_post_match/nicegui_dir/new_code4/all_data2.json', 'w') as f:
    json.dump(data, f, indent=4)
```
